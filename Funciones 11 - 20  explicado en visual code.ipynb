## Funciones matemáticas y de agregación

### 11. `pow()`
**¿Qué es?** Eleva un número a una potencia. Opcionalmente calcula el módulo de esa potencia.

**Sintaxis:** `pow(base, exponente, modulo=None)`

```python
print(pow(2, 3))     # 8
print(pow(5, 2))     # 25
print(pow(9, 0.5))   # 3.0 (raíz cuadrada)
print(pow(2, 5, 7))  # (2**5) % 7 = 4
```

### 12. `divmod()`
**¿Qué es?** Devuelve una tupla con el cociente y el residuo de una división entera.

**Sintaxis:** `divmod(a, b)`

```python
print(divmod(17, 5))  # (3, 2)
cociente, residuo = divmod(100, 7)
print(f"Cociente: {cociente}, Residuo: {residuo}")

# Ejemplo práctico: 135 minutos a horas y minutos
horas, minutos = divmod(135, 60)
print(f"{horas} h y {minutos} min")
```

### 13. `sum()`
**¿Qué es?** Suma todos los elementos numéricos de un iterable.

**Sintaxis:** `sum(iterable, inicio=0)`

```python
notas = [4.5, 3.8, 5.0, 4.2]
print(sum(notas))               # 17.5
print(sum(notas) / len(notas))  # promedio
print(sum([1, 2, 3], 10))       # 16
print(sum(range(1, 101)))       # 5050
```

### 14. `min()`
**¿Qué es?** Devuelve el elemento más pequeño de un iterable o de varios argumentos.

**Sintaxis:** `min(iterable, key=None, default=...)` o `min(a, b, c, ...)`

```python
print(min(8, 3, 15))            # 3
temperaturas = [18, 22, 9, 14]
print(min(temperaturas))        # 9
palabras = ["manzana", "pera", "uva"]
print(min(palabras, key=len))   # uva
print(min([], default="vacía")) # vacía
```

### 15. `max()`
**¿Qué es?** Devuelve el elemento más grande de un iterable o de varios argumentos.

**Sintaxis:** `max(iterable, key=None, default=...)` o `max(a, b, c, ...)`

```python
print(max(8, 3, 15))            # 15
edades = [21, 35, 18, 42]
print(max(edades))              # 42
palabras = ["manzana", "pera", "uva"]
print(max(palabras, key=len))   # manzana
print(max([], default=0))       # 0
```

## Funciones para organizar y recorrer colecciones

### 16. `reversed()`
**¿Qué es?** Devuelve un iterador que recorre una secuencia en orden inverso, sin modificar el original.

**Sintaxis:** `reversed(secuencia)`

```python
numeros = [1, 2, 3, 4]
print(list(reversed(numeros)))      # [4, 3, 2, 1]
print(numeros)                      # el original no cambia
print("".join(reversed("Python")))  # nohtyP
for n in reversed(range(1, 4)):
    print(n)
```

### 17. `range()`
**¿Qué es?** Genera una secuencia de enteros, muy usada en ciclos `for`.

**Sintaxis:** `range(fin)`, `range(inicio, fin)` o `range(inicio, fin, paso)`. El valor final no se incluye.

```python
print(list(range(5)))         # [0, 1, 2, 3, 4]
print(list(range(2, 8)))      # [2, 3, 4, 5, 6, 7]
print(list(range(0, 20, 5)))  # [0, 5, 10, 15]
print(list(range(10, 0, -3))) # [10, 7, 4, 1]
for i in range(1, 4):
    print("Vuelta", i)
```

### 18. `list()`
**¿Qué es?** Crea una lista, vacía o a partir de cualquier iterable.

**Sintaxis:** `list(iterable)`

```python
print(list())              # []
print(list("Hola"))        # ['H', 'o', 'l', 'a']
print(list((1, 2, 3)))     # [1, 2, 3]
print(list(range(3)))      # [0, 1, 2]
print(list({"a": 1, "b": 2}))  # ['a', 'b'] (las claves)
```

### 19. `tuple()`
**¿Qué es?** Crea una tupla (colección ordenada e **inmutable**) a partir de un iterable.

**Sintaxis:** `tuple(iterable)`

```python
print(tuple([1, 2, 3]))   # (1, 2, 3)
print(tuple("abc"))       # ('a', 'b', 'c')
print(tuple(range(4)))    # (0, 1, 2, 3)
coordenadas = tuple([4.6, -74.1])
print(coordenadas)
try:
    coordenadas[0] = 5
except TypeError as e:
    print("Error:", e)  # las tuplas no se pueden modificar
```

### 20. `sorted()`
**¿Qué es?** Devuelve una **nueva lista** ordenada con los elementos de un iterable, sin modificar el original.

**Sintaxis:** `sorted(iterable, key=None, reverse=False)`

```python
numeros = [5, 2, 9, 1]
print(sorted(numeros))                # [1, 2, 5, 9]
print(sorted(numeros, reverse=True))  # [9, 5, 2, 1]
print(numeros)                        # el original no cambia
palabras = ["uva", "manzana", "pera"]
print(sorted(palabras, key=len))      # por longitud
print(sorted("python"))               # letras ordenadas
```
